```python
import pandas as pd
import numpy as np
data = {
    'Weather': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain', 'Overcast', 'Sunny', 'Sunny', 'Rain', 'Sunny', 'Overcast', 'Overcast', 'Rain'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Cool', 'Mild', 'Cool', 'Mild', 'Mild', 'Mild', 'Hot', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'High'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Strong'],
    'PlayTennis': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}
df = pd.DataFrame(data)
def prior_probability(df, target_column):
    prior = df[target_column].value_counts(normalize=True)
    return prior
def likelihood(df, feature_column, feature_value, target_column, target_value):
    feature_target_df = df[df[target_column] == target_value]
    likelihood = len(feature_target_df[feature_target_df[feature_column] == feature_value]) / len(feature_target_df)
    return likelihood
def posterior_probability(df, new_instance, target_column):
    target_values = df[target_column].unique()
    posteriors = {}
    for target_value in target_values:
        prior = prior_probability(df, target_column)[target_value]
        likelihoods = 1
        for feature in new_instance:
            likelihoods *= likelihood(df, feature, new_instance[feature], target_column, target_value)
        posteriors[target_value] = prior * likelihoods
    return posteriors
new_instance = {'Weather': 'Sunny', 'Temperature': 'Cool', 'Humidity': 'High', 'Wind': 'Strong'}
posteriors = posterior_probability(df, new_instance, 'PlayTennis')
print("Posterior probabilities:", posteriors)
prediction = max(posteriors, key=posteriors.get)
print("Predicted class:", prediction)
```
