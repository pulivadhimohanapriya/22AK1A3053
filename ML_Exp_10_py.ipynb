```python
import pandas as pd
data = {
    'Weather': ['Sunny', 'Sunny', 'Overcast', 'Rainy', 'Rainy', 'Rainy', 'Overcast', 'Sunny', 'Sunny', 'Rainy', 'Sunny', 'Overcast', 'Overcast', 'Rainy'],
    'Playable': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}
df = pd.DataFrame(data)
frequency_table = pd.crosstab(df['Weather'], df['Playable'])
print("Frequency Table:")
print(frequency_table)
total_playable = df['Playable'].value_counts()
likelihood_table = frequency_table.div(total_playable, axis=1)
print("\nLikelihood Table (Conditional Probabilities):")
print(likelihood_table)
prior_prob_playable = total_playable / len(df)
print("\nPrior Probabilities:")
print(prior_prob_playable)
weather_condition = 'Sunny'
likelihood_yes = likelihood_table.loc[weather_condition, 'Yes']
likelihood_no = likelihood_table.loc[weather_condition, 'No']
posterior_yes = likelihood_yes * prior_prob_playable['Yes']
posterior_no = likelihood_no * prior_prob_playable['No']
total_posterior = posterior_yes + posterior_no
posterior_yes /= total_posterior
posterior_no /= total_posterior
print(f"\nPosterior Probability for 'Yes' given Weather = {weather_condition}: {posterior_yes}")
print(f"Posterior Probability for 'No' given Weather = {weather_condition}: {posterior_no}")
predicted_class = 'Yes' if posterior_yes > posterior_no else 'No'
print(f"\nPredicted outcome for Weather = {weather_condition}: {predicted_class}")
```
