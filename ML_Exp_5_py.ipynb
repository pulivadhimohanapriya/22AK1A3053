```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
import copy

# Load dataset
dataset = pd.read_csv('tennis.csv')
X = dataset.iloc[:, 1:].values
attribute = ['outlook', 'temp', 'humidity', 'wind']

# Node class to represent the decision tree
class Node(object):
    def __init__(self):
        self.value = None
        self.decision = None
        self.childs = []

# Function to calculate entropy
def findEntropy(data, rows):
    yes = 0
    no = 0
    idx = len(data[0]) - 1
    for i in rows:
        if data[i][idx] == 'Yes':
            yes += 1
        else:
            no += 1
    x = yes / (yes + no)
    y = no / (yes + no)
    entropy = 0
    if x != 0 and y != 0:
        entropy = -1 * (x * math.log2(x) + y * math.log2(y))
    ans = -1
    if x == 1:
        ans = 1
    if y == 1:
        ans = 0
    return entropy, ans

# Function to calculate the maximum information gain
def findMaxGain(data, rows, columns):
    maxGain = 0
    retidx = -1
    entropy, ans = findEntropy(data, rows)
    if entropy == 0:
        return maxGain, retidx, ans
    for j in columns:
        mydict = {}
        for i in rows:
            key = data[i][j]
            if key not in mydict:
                mydict[key] = 1
            else:
                mydict[key] += 1
        gain = entropy
        for key in mydict:
            yes = 0
            no = 0
            for k in rows:
                if data[k][j] == key:
                    if data[k][-1] == 'Yes':
                        yes += 1
                    else:
                        no += 1
            x = yes / (yes + no)
            y = no / (yes + no)
            if x != 0 and y != 0:
                gain -= (mydict[key] / len(rows)) * (x * math.log2(x) + y * math.log2(y))
        if gain > maxGain:
            maxGain = gain
            retidx = j
    return maxGain, retidx, ans

# Function to build the decision tree
def buildTree(data, rows, columns):
    maxGain, idx, ans = findMaxGain(data, rows, columns)
    root = Node()
    if maxGain == 0:
        if ans == 1:
            root.value = 'Yes'
        else:
            root.value = 'No'
        return root
    root.value = attribute[idx]
    mydict = {}
    for i in rows:
        key = data[i][idx]
        if key not in mydict:
            mydict[key] = 1
        else:
            mydict[key] += 1
    newcolumns = copy.deepcopy(columns)
    newcolumns.remove(idx)
    for key in mydict:
        newrows = [i for i in rows if data[i][idx] == key]
        temp = buildTree(data, newrows, newcolumns)
        temp.decision = key
        root.childs.append(temp)
    return root

# Function to traverse the tree and print the nodes
def traverse(root):
    if root.decision is not None:
        print(f"Decision: {root.decision}")
    print(f"Value: {root.value}")
    for child in root.childs:
        traverse(child)

# Main function to calculate and print the decision tree
def calculate():
    rows = [i for i in range(len(X))]
    columns = [i for i in range(len(attribute))]
    root = buildTree(X, rows, columns)
    root.decision = 'Start'
    traverse(root)

# Call the main function
calculate()
```
